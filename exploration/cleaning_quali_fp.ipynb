```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
results = pd.read_csv("cleaned_data/f1db-races-race-results-CLEANED.csv")
```

```python
#get (raceId, driverId) pairs for all actual race starts in cleaned race results
starts = list(zip(results.raceId.values, results.driverId.values))
starts
```

```python
fp1 = pd.read_csv("data_f1db/f1db-races-free-practice-1-results.csv")
fp2 = pd.read_csv("data_f1db/f1db-races-free-practice-2-results.csv")
fp3 = pd.read_csv("data_f1db/f1db-races-free-practice-3-results.csv")
fp4 = pd.read_csv("data_f1db/f1db-races-free-practice-4-results.csv")

quali = pd.read_csv("data_f1db/f1db-races-qualifying-results.csv")
```

```python
df = fp1

fp1 = df[pd.Series(list(zip(df.raceId,df.driverId)), index=df.index).isin(starts)]
```

```python
df = fp2

fp2 = df[pd.Series(list(zip(df.raceId,df.driverId)), index=df.index).isin(starts)]
```

```python
df = fp3

fp3 = df[pd.Series(list(zip(df.raceId,df.driverId)), index=df.index).isin(starts)]
```

```python
df = fp4

fp4 = df[pd.Series(list(zip(df.raceId,df.driverId)), index=df.index).isin(starts)]
```

```python
df = quali

quali = df[pd.Series(list(zip(df.raceId,df.driverId)), index=df.index).isin(starts)]
```

```python
fp1.to_csv("cleaned_data/f1db-races-free-practice-1-results-CLEANED.csv")
fp2.to_csv("cleaned_data/f1db-races-free-practice-2-results-CLEANED.csv")
fp3.to_csv("cleaned_data/f1db-races-free-practice-3-results-CLEANED.csv")
fp4.to_csv("cleaned_data/f1db-races-free-practice-4-results-CLEANED.csv")

quali.to_csv("cleaned_data/f1db_races-qualifying-results-CLEANED.csv")
```

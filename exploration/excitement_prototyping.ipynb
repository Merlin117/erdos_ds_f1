```python
races = pd.read_csv("data_f1db/f1db-races.csv")
races = races[(races.year >= 1986) & (races.year < 2025) & (races.sprintQualifyingFormat.isnull())]
```

```python

for id in races['id']:
    race = race_results[(race_results.raceId == id) & ~(race_results.positionText.isin(["DNQ","DNS","EX","DNPQ","DNP"]) | ((race_results.positionText == "DSQ") & race_results.gridPositionNumber.isnull()))]
    if len(race.index) == 0:
        print(id)
        continue
    if id == 426:
        print(sum(abs(race['positionDisplayOrder'] - race['gridPositionNumber'])))
    races.loc[races.id==id,'total_position_changes'] = sum(abs(race['positionDisplayOrder'] - race['gridPositionNumber']))
    races.loc[races.id==id,'avg_position_changes'] = races[races.id==id]['total_position_changes'] / len(race.index)
```

```python
sns.scatterplot(data=races,x="id",y="avg_position_changes")
```
